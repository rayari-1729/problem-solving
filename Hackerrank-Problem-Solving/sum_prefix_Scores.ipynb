```python
# HashTabke BruteForce
words = ["abc","ab","bc","b"]
from collections import defaultdict
count =defaultdict(int)
for i in words:
  for j in range(len(i)):
      count[i[:j+1]] += 1
res = []
for i in words:
  value = 0
  for j in range(len(i)):
    value += count[i[:j+1]]
  res.append(value)

print(res)









```
