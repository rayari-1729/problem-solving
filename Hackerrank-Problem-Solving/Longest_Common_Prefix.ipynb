```python
arr1 = [1,10,100]
arr2 = [10000]

def longestCommonPrefix( arr1, arr2) -> int:
        # Idea is use hash map
        if len(arr1) > len(arr2):
            arr1 , arr2 = arr2, arr1
        prefix_set = set()
        for n in arr1:
            while n and n not in prefix_set : #(last part is the optimization) 1234, 12345
                prefix_set.add(n)
                n= n//10

        res = 0

        for n in arr2:

            while n and n not in prefix_set:
                n = n//10
            if n:
                res = max(res, len(str(n)))

        return res

longestCommonPrefix(arr1,arr2)
```
